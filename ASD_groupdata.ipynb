```python
import glob
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
```

```python
start_row = 1
```

```python
#EXAMPLE
# collect all sample files in the group folder (adjust pattern if needed)
files = sorted(glob.glob('Group data/ASD/Group 2/MM_350000*.asd.txt'))

# read each file into a dataframe indexed by Wavelength
dfs = []
for f in files:
    name = os.path.basename(f)
    df = pd.read_csv(f, sep='\t', names=['Wavelength', name], skiprows=start_row)
    df = df.set_index('Wavelength')
    dfs.append(df)
# combine into one DataFrame aligned on Wavelength
alldata = pd.concat(dfs, axis=1)
# compute average reflectance across samples (skip NaNs)
alldata['Avg refl'] = alldata.mean(axis=1)

# example: plot average spectrum
wl = alldata.index.astype(float)
refl_avg = alldata['Avg refl']

# fig = plt.figure(figsize=(14,6))
plt.plot(wl, refl_avg, label='Average')
plt.xlabel('Wavelength (nm)')
plt.ylabel('Reflectance')
plt.grid()
plt.title('Average spectrum — MM_350000')
plt.ylim(0,1)
plt.legend()
plt.show()
```
